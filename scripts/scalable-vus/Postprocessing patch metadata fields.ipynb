```python
# Check that all genes have phenotypes
or_df['Gene'].isin(gene_phenotypes).all()
```

```python
or_df['Case inclusion phenotypes'] = or_df['Gene'].map(lambda g: list(gene_phenotypes[g][0]))
or_df['Control exclusion phenotypes'] = or_df['Gene'].map(lambda g: list(gene_phenotypes[g][1]))
or_df
```

```python
# ENSG map
ensg_df = pd.read_table(
    '~/workspace/migrated-workspace/aux_data/gene_metadata.txt',
    usecols=['Gene stable ID version', 'Chromosome/scaffold name', 'Gene name'],
    dtype=str
).drop_duplicates()
```

```python
# Filter out to chromosomes only and our genes only
ensg_df = ensg_df[
    ensg_df['Gene name'].isin(or_df['Gene']) &
    ~ensg_df['Chromosome/scaffold name'].str.endswith('PATCH')
].set_index('Gene name')
```

```python
or_df['ENSG'] = or_df['Gene'].map(ensg_df['Gene stable ID version'])
```
