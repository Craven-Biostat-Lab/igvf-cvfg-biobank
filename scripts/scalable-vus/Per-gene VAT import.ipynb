## Load VAT in Hail

```python
import os
from pathlib import Path

output_path = Path.home() / 'workspace/scalable-vus-validation/vat'
output_dir = output_path.as_posix()
vat_path = 'gs://vwb-aou-datasets-controlled/v8/wgs/short_read/snpindel/aux/vat/vat_complete.bgz.tsv.gz'
#vat_path = (
#    Path.home() / 'workspace' /
#    'vwb-aou-datasets-controlled-v8r9/v8/wgs/short_read/snpindel/aux/vat/vat_complete.bgz.tsv.gz'
#).as_posix()
```

```python
output_path.mkdir(parents=True, exist_ok=True)
```

```python
import hail as hl
hl.init()
```

```python
vat_table = hl.import_table(vat_path, force_bgz=True, types={'position': hl.tint32})
```

## Define gene list

```python
genes_of_interest = [
    'BAP1',
    'BARD1',
    'BRCA1',
    'BRCA2',
    'CHEK2',
    'GCK',
    'KCNE1',
    'KCNH2',
    'KCNQ4',
    'LDLR',
    'MSH2',
    'PALB2',
    'PTEN',
    'RAD51C',
    'RAD51D',
    'SCN5A',
    'TP53',
    'TSC2'
]

# Remove genes for which we have previously extracted VATs
# genes_for_extraction = [
#    gene for gene in genes_of_interest
#    if gene not in {'BRCA1', 'KCNH2', 'CHEK2', 'CALM1', 'SCN5A'}
#]
```

## Extract specific gene VATs

```python
for gene in genes_of_interest:
    gene_vat = vat_table.filter(vat_table.gene_symbol == gene)
    gene_vat.export(f'{output_dir}/{gene}_vat.tsv')
```
